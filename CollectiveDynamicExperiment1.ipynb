This project investigates whether collectively generated, embodied synchrony induces neuroplasticity and shapes learning, attachment, resiliency, and altered states of consciousness across the lifespan. I plan to use wireless wearable biosensors such as EEG, PPG (3-wavelength) to derive heart rate, heart rate variability, respiration, oxygen saturation, EDA / GSR (electrodermal activity / galvanic skin response), 9-axis IMU (accel, gyro, mag) to derive movements, activity, gesture, rotation, and cardinal direction, body temperature and 3D video capture of group body movement that drives responsively generated sound, with the option to make those sounds culturally specific in order to consider culture as a modifier of self-organized synchronous dynamics.
