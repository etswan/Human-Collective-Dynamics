# Summary

This project aims to model the transitions in a closed-loop feedback system of human to human and human to audio coupled oscillations, with group movement-generated music serving as a real-time output of the collective state alongside heart rate, and both limb and agent center locations. When and how does the collective shift between synchrony states, and do those shifts track markers of altered states of consciousness (HR) with persistence (psych instrument)? One aim is mechanistic specificity of transitions and synchrony. I will also estimate how near a transition the system is and whether groups actively regulate that distance. At a longer timescale, the functional question is whether collective synchrony carries consequences for altered states of consciousness and wellbeing.
