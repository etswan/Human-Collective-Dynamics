# Summary

This project investigates whether collectively generated, embodied synchrony induces neuroplasticity and shapes learning, attachment, resiliency, and altered states of consciousness across the lifespan. I plan to use wireless wearable biosensors such as EEG, PPG (3-wavelength)/ECG to derive heart rate, heart rate variability, respiration, oxygen saturation, EDA / GSR (electrodermal activity / galvanic skin response), 9-axis IMU (accel, gyro, mag) to derive movements, activity, gesture, rotation, and cardinal direction, body temperature and 3D video capture of group body movement that drives responsively generated sound, with the option to make those sounds culturally specific in order to consider culture as a modifier of self-organized synchronous dynamics.

**Update:** Instead of focusing on describing and mapping states, I've shifted to the transitions in the closed-loop feedback system. The setup is essentially a large coupled oscillator: people and sound mutually driving each other, human-human and human-audio. The music is treated as a readout of the collective movement via an additional channel alongside the EEG, heart rate, and location. What I'm most interested in are the transitions themselves, when and how the collective shifts between states, and whether those shifts track markers of altered states of consciousness.

# Current Plan: 25 September

**Signal Prep/Reduction**

- CHARM manifold trajectory (criticality-sensitive, non-local structure). Deco, Sanz Perl & Kringelbach 2025. 
- Coupling signals in parallel: Kuramoto (group, EEG phase) + PLV (pairwise); cross-modal coupling (windowed cross-correlation / MI / coherence) for heart rate, location, movement-to-music. Reinero, Dikker & Van Bavel 2021.

**Detect & Segment Transitions**

- HMM on the reduced signals: discrete states, transition timing and rates. Stevner et al. 2019.
- Check BIC/AIC or held-out for fit/significance of states

**Transition Analysis**

- SDE / energy-landscape estimation on the reduced trajectory: recover drift, wells, barriers. Interpret as states of the coupled human-human / human-audio system. Kang, Pae & Park 2019.
- Graph-theoretical view? Kang, Pae & Park 2019

**Predictive Model**

- HMM (but not of continuous driving dynamics/memory), validate on held-out sessions. (Take the manifold coordinates from CHARM and add time delay before SDE?)
- SDE, or a coupled-oscillator/Hopf model – wells & barriers, run on CHARM (dimensionality reduction), not raw channels to better catch drift, estimate in windows to not over generalize (assumes a stationary landscape) – validate wells against HMM states. Match sampling rate to timescale of transitions.

**Altered States of Consciousness (ASC)**

- HR/HRV Autonomic within-subject sympathetic-parasympathetic co-activation
- EEG Neural complexity/entropy markers
- Pre & post-test instrument (11-D ASC / MODTAS or similar)
